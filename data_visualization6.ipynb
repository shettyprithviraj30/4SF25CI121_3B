```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("amazon_prime_titles.csv")

df["duration_value"] = pd.to_numeric(df["duration"].str.extract(r"(\d+)")[0])

df.head()
```

1. Line graph - number of titles released per year

```python
titles_per_year = df[df["release_year"] >= 1980].groupby("release_year").size()

plt.figure(figsize=(10, 5))
plt.plot(titles_per_year.index, titles_per_year.values, color="teal", marker="o", markersize=4)
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.title("Number of Titles by Release Year (1980 onwards)")
plt.grid(linestyle="--", alpha=0.4)
plt.show()
```

2. Bar graph - top 10 genres

```python
genre_counts = df["listed_in"].str.split(", ").explode().value_counts().head(10)

plt.figure(figsize=(11, 5))
plt.bar(genre_counts.index, genre_counts.values, color="slateblue")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.title("Top 10 Genres on Amazon Prime")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

3. Histogram - distribution of movie runtime

```python
movies = df[df["type"] == "Movie"]

plt.figure(figsize=(8, 5))
plt.hist(movies["duration_value"], bins=20, color="orange", edgecolor="black")
plt.xlabel("Runtime (minutes)")
plt.ylabel("Number of Movies")
plt.title("Distribution of Movie Runtime")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.show()
```

4. Scatter plot - release year vs movie runtime

```python
plt.figure(figsize=(9, 5))
plt.scatter(movies["release_year"], movies["duration_value"], color="crimson", alpha=0.3, s=12)
plt.xlabel("Release Year")
plt.ylabel("Runtime (minutes)")
plt.title("Release Year vs Movie Runtime")
plt.grid(linestyle="--", alpha=0.4)
plt.show()
```

5. Pie chart - Movies vs TV Shows

```python
type_counts = df["type"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    type_counts.values,
    labels=type_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["gold", "red"],
    explode=(0.05, 0),
    shadow=True
)
plt.title("Movies vs TV Shows")
plt.show()
```

Extra - top age ratings (bar) and TV show seasons

```python
rating_counts = df["rating"].value_counts().head(8)

rating_counts.plot(kind="bar", color="teal", figsize=(9, 5))
plt.xlabel("Age Rating")
plt.ylabel("Number of Titles")
plt.title("Top Age Ratings")
plt.xticks(rotation=0)
plt.show()
```

```python
tv = df[df["type"] == "TV Show"]
seasons = tv["duration_value"].value_counts().sort_index().head(10)

plt.bar(seasons.index.astype(int).astype(str), seasons.values, color="purple")
plt.xlabel("Number of Seasons")
plt.ylabel("Number of TV Shows")
plt.title("TV Shows by Number of Seasons")
plt.show()
```
